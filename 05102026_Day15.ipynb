```python
# import pandas and matplotlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("/content/StudentsPerformance.csv", nrows=50)
data.head()

data.hist()

# show plot
plt.show()
```

```python
data.plot.box()

# individual attribute box plot
plt.boxplot(data)
plt.show()
```

```python
x = data['math score']
y = data['reading score']
z=data['writing score']
plt.plot(x, y, z)
plt.show()
```

```python
x = data['math score']
y = data['reading score']
z = data['writing score']

plt.scatter(x, y, c=z, cmap='viridis', alpha=0.7)
plt.colorbar(label='Writing Score')

plt.xlabel('Math Score')
plt.ylabel('Reading Score')
plt.show()
```

```python
x = data['math score']
y = data['reading score']
z = data['writing score']

# Plots reading score (y) and writing score (z) side-by-side using math score (x) as the base
plt.bar(x, y, width=0.4, label='Reading Score')
plt.bar(x + 0.4, z, width=0.4, label='Writing Score')

plt.legend()
plt.show()
```
